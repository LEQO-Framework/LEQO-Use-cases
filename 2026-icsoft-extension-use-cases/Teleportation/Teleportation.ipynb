# Teleportation Use Case

```python
from collections import Counter
from pathlib import Path

from qiskit import transpile
from qiskit_aer import AerSimulator

PROGRAM_FILE = Path("teleportation.py")
SHOTS = 4096
```

```python
program = PROGRAM_FILE.read_text(encoding="utf-8")
print(f"{PROGRAM_FILE}: {len(program.splitlines())} lines")
```

```python
namespace = {}
exec(program.split("' --- EXECUTION ---'")[0], namespace)
qc = namespace["qc"]
simulator = AerSimulator()
counts = simulator.run(transpile(qc, simulator), shots=SHOTS).result().get_counts()

role = {0: "m0", 1: "m1", 2: "bob"}  # qubit 0: message, 1: Alice's and 2: Bob's half of the Bell pair
clbit = {role[qc.find_bit(item.qubits[0]).index]: qc.find_bit(item.clbits[0]).index
         for item in qc.data if item.operation.name == "measure"}
outcomes = Counter()
for word, shots in counts.items():
    bits = word.replace(" ", "")[::-1]
    outcomes[bits[clbit["m0"]], bits[clbit["m1"]], bits[clbit["bob"]]] += shots

print("m0 m1 bob  shots")
for (m0, m1, bob), shots in sorted(outcomes.items()):
    print(f"{m0}  {m1}  {bob}    {shots}")
```
